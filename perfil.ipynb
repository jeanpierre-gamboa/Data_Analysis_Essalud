```python
df["DES_SER"] = df["DES_SER"].str.strip()
df.groupby("DES_SER")["CANTIDAD"].sum().sort_values(ascending=False)
```

```python
df["DES_ACT"] = df["DES_ACT"].str.strip()
df.groupby("DES_ACT")["CANTIDAD"].sum().sort_values(ascending=False)
```

```python
df[df["DES_SER"] != ""].groupby("DES_ACT")["CANTIDAD"].sum().sort_values(ascending=False)
```

```python
df[df["DES_ACT"] == "Hras.Efec.(Eje)"].groupby("DES_SER")["CANTIDAD"].sum().sort_values(ascending=False)
```

```python
df[df["DES_ACT"] == "Horas Prog."].groupby("DES_SER")["CANTIDAD"].sum().sort_values(ascending=False)
```

```python
df_horas = df[df["DES_ACT"].isin(["Horas Prog.", "Hras.Efec.(Eje)"])]
df_tabla = df_horas.groupby(["DES_SER","DES_ACT"])["CANTIDAD"].sum()
df_tabla = df_tabla.unstack()
df_tabla["hr_no_ejecutada"] = df_tabla["Horas Prog."] - df_tabla["Hras.Efec.(Eje)"]
df_tabla["tasa_no_cumplimiento"] = (df_tabla["hr_no_ejecutada"] / df_tabla["Horas Prog."])*100
df_tabla.sort_values(by="tasa_no_cumplimiento", ascending= False)
```
