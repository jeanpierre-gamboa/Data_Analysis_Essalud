```python
df["DES_SER"] = df["DES_SER"].str.strip()
df.groupby("DES_SER")["CANTIDAD"].sum().sort_values(ascending=False)
```

```python
df["DES_ACT"] = df["DES_ACT"].str.strip()
df.groupby("DES_ACT")["CANTIDAD"].sum().sort_values(ascending=False)
```

```python
df_horas = df[df["DES_ACT"].isin(["Horas Prog.", "Hras.Efec.(Eje)"])]
df_tabla = df_horas.groupby(["DES_SER","DES_ACT"])["CANTIDAD"].sum()
df_tabla = df_tabla.unstack()
df_tabla["hr_no_ejecutada"] = df_tabla["Horas Prog."] - df_tabla["Hras.Efec.(Eje)"]
df_tabla["tasa_no_cumplimiento"] = (df_tabla["hr_no_ejecutada"] / df_tabla["Horas Prog."])*100
df_tabla.sort_values(by="tasa_no_cumplimiento", ascending= False)
```

```python
df_horas = df[df["DES_ACT"].isin(["Horas Prog.", "Hras.Efec.(Eje)"])]
df_tabla_dep = df_horas.groupby(["DEPARTAMENTO","DES_ACT"])["CANTIDAD"].sum()
df_tabla_dep = df_tabla_dep.unstack()
df_tabla_dep["hr_no_ejecutada"] = df_tabla_dep["Horas Prog."] - df_tabla_dep["Hras.Efec.(Eje)"]
df_tabla_dep["tasa_no_cumplimiento"] = (df_tabla_dep["hr_no_ejecutada"] / df_tabla_dep["Horas Prog."])*100
df_tabla_dep.sort_values(by="tasa_no_cumplimiento", ascending= False)
```

```python
df["PERIODO"].unique()
```

```python
deps = [
    "AMAZONAS", "LAMBAYEQUE", "AYACUCHO", "MADRE DE DIOS", "CAJAMARCA",
    "UCAYALI", "LIMA", "CALLAO", "HUANUCO", "AREQUIPA",
    "PASCO", "MOQUEGUA", "PIURA", "LA LIBERTAD", "LORETO",
]
df_horas = df[df["DES_ACT"].isin(["Horas Prog.", "Hras.Efec.(Eje)"])]
df_horas_filt = df_horas[df_horas["DEPARTAMENTO"].isin(deps)]
df_horas_filt["PERIODO_AÑO"] = (df_horas_filt["PERIODO"]//100)
df_tabla_per = df_horas_filt.groupby(["DEPARTAMENTO","PERIODO_AÑO","DES_ACT"])["CANTIDAD"].sum()
df_tabla_per = df_tabla_per.unstack()
df_tabla_per["hr_no_ejecutada"] = df_tabla_per["Horas Prog."] - df_tabla_per["Hras.Efec.(Eje)"]
df_tabla_per["tasa_no_cumplimiento"] = (df_tabla_per["hr_no_ejecutada"] / df_tabla_per["Horas Prog."])*100
df_tabla_per
```

```python
pos_alta = df_tabla_per.groupby("DEPARTAMENTO")["tasa_no_cumplimiento"].idxmax()
df_tabla_per.loc[pos_alta].sort_values(by= "tasa_no_cumplimiento", ascending=False)
```

```python
pos_baja = df_tabla_per.groupby("DEPARTAMENTO")["tasa_no_cumplimiento"].idxmin()
df_tabla_per.loc[pos_baja].sort_values(by= "tasa_no_cumplimiento", ascending=False)
```
